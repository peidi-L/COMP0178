## 2. How database-design notes should prepare you

- Explain primary/candidate/composite keys, foreign keys, and functional dependencies with examples.
- For normalization, state assumptions, derive dependencies, identify the violation, and show a corrected schema.
- **2023 Q1:** inspect listOfStudents for a repeating/multivalued field and reason from the stated class/module/teacher assumptions; do not infer dependencies solely from column names.
- **2024 Q1:** the diagram has Class-Student many-to-many, Class-Classroom many-to-many with dayOfWeek/time relationship attributes, and Teacher-Class one-to-many. Practice junction tables, relationship attributes, and placing the teacher FK on Class.
- Scheduling relationship keys need stated assumptions if one class/room pair can meet more than once; a diagram does not automatically settle every uniqueness rule.
- **2025 Q1:** practice extracting User, Restaurant, Review, List, and list membership from prose, with relationship multiplicities and public/private list information.
- Future design lessons should include one prose-to-ER task and one ER-to-schema or normalization task, when relevant.

## 3. Query-processing preparation: a connected chain

- **Step 1:** state a meaningful business question and write SQL using the required number of tables.
- **Step 2:** translate the same semantics into relational algebra, including join predicates and projected attributes.
- **Step 3:** draw a tree matching the algebra exactly.
- **Step 4:** draw a different but equivalent tree, such as pushing a single-table selection below a join when valid.
- **Step 5:** annotate estimated row counts and calculate disk costs under explicit assumptions.
- Retain keys and predicate columns when pushing projection down; otherwise later joins may become impossible.
- SQL can preserve duplicates; classical relational-algebra projection removes them. State DISTINCT/set assumptions for comparable results.
- Use extended relational algebra if translating aggregate queries; alternatively choose a clear nonaggregate practice query initially.
- 2023's schema is a real-estate agency; 2024 uses Appointments, Patients, Physicians; 2025 uses Customers, Items, LineItems, Orders.
- **2025 Q2(a) explicitly requires a different question from its suggested quantity-by-state example.**
- Cost assumptions across the papers include no indexes, one tuple per read, base tables fitting in memory, intermediate results stored on disk, and final results returned rather than stored.
- Do not assume repeated base-table reads for every pair if the stated memory model permits retaining tables. State materialization/read/write choices consistently.
- 2023 asks for disk reads; 2024/2025 say disk accesses. Explain whether counts include writes and show separate read/write contributions where relevant.

## 4. Transaction notes should show traces and graphs

- Conflict serializability: identify operations in different transactions on the same item with at least one write; create directed precedence edges; test for cycles; give a topological serial order if possible.
- Deadlock: identify who holds each lock and who waits for whom; show a feasible waiting cycle rather than merely saying opposite lock order is risky.
- Timestamp ordering: trace transaction timestamps and each item's read/write timestamps; identify the first rejected operation under the stated protocol.
- Recoverability: track which transaction reads another's write and compare their commit order. Distinguish recoverable, cascadeless, and strict schedules.
- Recovery: explain immediate versus deferred update and which undo/redo operations are required under the relevant assumptions.
- Include short-answer versions matching the exam's 1-2 sentence requests, followed by fuller teaching explanations.

## 5. Distributed and non-relational notes

- Distinguish **fragmentation** (split data), **allocation** (place fragments), and **replication** (store copies).
- Justify choices using query locality, cross-site traffic, availability, consistency, and workload—not only naming techniques.
- Distributed counts: compare shipping matching records with computing local counts and summing small results.
- Distributed averages: combine local totals and appropriate population counts; do not simply average site averages unless denominator populations are equal.
- For 2025's posts-per-user average, state whether users with zero posts count and what user population defines the denominator.
- User-ID routing: explain how the allocation/partition function identifies the owning site and whether user-related fragments are co-located.
- 2PC recovery: reason from durable log records and the protocol's decision rather than choosing a new outcome after recovery.
- Compare relational and semistructured querying, schema flexibility, integrity, and trade-offs with concrete examples.
- Include XML navigation and XPath/FLWOR practice for 2025 Q5; note malformed example markup if relevant rather than silently assuming it is valid XML.
- Explain how relational and semistructured data can coexist, and how hashing supports partitioning, access structures, or hash joins.
