## 2. HTML: content and structure - pages 3-5

**HTML = Hypertext Markup Language.** Tags mark elements such as headings, paragraphs, links, and forms. HTML is markup, rather than a general-purpose programming language.

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Book reviews</title>
</head>
<body>
    <h1>Book reviews</h1>
    <p>Read a <a href="reviews.php">review</a>.</p>
</body>
</html>
```

- `<!DOCTYPE html>` selects HTML standards mode.
- `<html>` is the document's root element.
- `<head>` contains metadata and resource references; `<title>` names the browser tab.
- `<body>` contains the page content.
- `<h1>` is a heading; `<p>` is a paragraph.
- `<a>` is a link; its `href` **attribute** gives the destination.
- Most elements have opening and closing tags; elements such as `<input>` do not need a closing tag in HTML.

HTML and XML both use markup, but their parsing rules differ. The lecture's history credits Tim Berners-Lee and mentions W3C; this is background rather than a list to memorise.

## 3. HTTP and HTTPS - page 6

**HTTP = Hypertext Transfer Protocol.** A browser sends a **request** and the server returns a **response**.

```text
Request:  GET /reviews.php?bookId=12
Response: status + headers + body (for example, an HTML page)
```

- **Stateless:** HTTP does not automatically remember application state between requests.
- A website can add continuity using cookies and server-side sessions, for example to recognise a logged-in user.
- **HTTPS** protects HTTP communication using TLS.

**Clarification:** the slide mentions SSL alongside TLS. SSL is the obsolete predecessor; understand HTTPS as HTTP protected by TLS. Encryption protects communication, not the correctness of application logic.

## 4. Forms and submitting values - pages 7-8 and 23

A form collects values and sends them to a server endpoint.

```html
<h2>Create a new user account</h2>
<form method="post" action="createNewUser.php" name="newUserForm">
    <label for="firstName">First name:</label>
    <input id="firstName" type="text" name="firstName" size="30" required>

    <label for="familyName">Family name:</label>
    <input id="familyName" type="text" name="familyName" size="30" required>

    <input type="submit" value="Create new user">
</form>
```

- `method` specifies how to submit: here POST.
- `action` specifies the destination. The page 8 form omits it, submitting to the current document's URL.
- Each input's `name` supplies the key used by the server, such as `firstName`.
- `id` identifies an element in the document; the label's `for` connects to that ID.
- The submit button submits the form.
- `required` helps browser-side validation; the server must still validate received values.

If the user enters Ada and Lovelace, the submitted fields are `firstName=Ada` and `familyName=Lovelace`. The HTML has no database insert by itself.

## 5. GET versus POST - page 9

| Property | GET form submission | POST form submission |
|---|---|---|
| Form values normally appear | URL query string | Request body |
| Typical use | Search, browsing, filtering | Creating/updating records or submitting actions |
| Example | `browse.php?viewid=20&sort=true` | Fields sent to `createNewUser.php` |
| PHP access | `$_GET` | `$_POST` for ordinary form-encoded bodies |

In the URL example, `viewid` has value `20`, and `sort` has value `true` **as a string**, not automatically a PHP Boolean.

**Important clarification:** POST is not encryption. HTTPS protects transport for both methods. URL parameters can appear in history/logs; sensitive information should not be put in URLs. Refreshing a POST response may prompt resubmission, but a refresh prompt is not the definition of POST.

A POST request can also have URL query parameters; the table describes ordinary form submission behaviour.

## 6. CSS: presentation - pages 10-12

**CSS = Cascading Style Sheets.** Rules select elements and set properties.

```css
h1 {
    font-family: Georgia, "Times New Roman", Times, serif;
    font-size: 30px;
    color: #00ff66;
}
p {
    font-weight: bold;
    line-height: 1.3em;
    border-style: solid;
}
```

`h1` is the **selector**. Inside braces, each declaration has `property: value;`. `#00ff66` is a colour, `px` is a CSS pixel unit, and `em` is relative to font size.

Three ways to include CSS:

1. Inline: `<h1 style="color: green;">Reviews</h1>`.
2. An internal `<style> ... </style>` block in `<head>`.
3. An external file linked in `<head>`: `<link rel="stylesheet" href="css/styles.css">`.

External stylesheets help share presentation across pages and keep it separate from content. When several rules apply, CSS's cascade and specificity determine the result; file order alone is not the whole rule.

## 7. Classes, IDs, and Bootstrap - pages 13-15


## 8. JavaScript and the DOM - pages 16-17

Browser JavaScript responds to events and changes the page. **JavaScript is not Java.**

The **Document Object Model (DOM)** is the browser's object/tree representation of the HTML document. JavaScript can find elements, modify text, add sections, or hide/show content.

```html
<p id="status">Not saved</p>
<button id="preview" type="button">Preview message</button>
<script>
    document.getElementById("preview").addEventListener("click", function () {
        document.getElementById("status").textContent = "Preview ready";
    });
</script>
```

Clicking the button changes the DOM text. It does not itself update a database or prove that anything was saved.

- jQuery is a JavaScript library that simplifies common browser operations.
- **Correction to slide terminology:** Node.js is a JavaScript runtime, not a browser library. JavaScript can run server-side, although the tutorial's examples focus on the browser.
- Browser scripts can be disabled or bypassed: server validation remains necessary.

## 9. PHP: server-side execution - pages 18-21

**PHP = PHP: Hypertext Preprocessor** (the slide abbreviates the expansion incorrectly). It is a server-side scripting language commonly used to generate dynamic responses.

```php
<!DOCTYPE html>
<html lang="en">
<head><title>Example</title></head>
<body>
    <h1>A title</h1>
    <?php
    $date = date("l, F d, Y H:i");
    echo "<p>Today is $date</p>";
    ?>
</body>
</html>
```

- PHP code goes between `<?php` and `?>` when mixed with HTML.
- Variables start with `$`, such as `$date`.
- Ordinary statements end with a semicolon.
- `echo` sends text into the response.
- A double-quoted PHP string can interpolate variables.
- The server's timezone configuration determines the date shown; `H:i` uses 24-hour hours/minutes.
- Save as a `.php` file and request it through a configured PHP server.

The browser receives generated HTML, not the executed PHP source. Viewing page source therefore shows the generated paragraph. A server misconfiguration can expose source, so this assumes PHP is configured correctly.

## 10. PHP functions - page 22

```php
<?php
function factorial($number) {
    if ($number == 0) {
        return 1;
    }
    return $number * factorial($number - 1);
}
echo factorial(5); // 120
?>
```

A function accepts an argument and returns a result. This recursive example calls itself with a smaller value until the **base case**, zero.

```text
factorial(3) = 3 * factorial(2)
             = 3 * 2 * factorial(1)
             = 3 * 2 * 1 * factorial(0)
             = 6
```

The example assumes a non-negative integer small enough for the numeric range. A negative input would never reach the zero base case. PHP syntax uses braces and semicolons, while the underlying function idea is similar to Python.

## 11. Reading form data in PHP - pages 23-24

The form from section 4 submits to `createNewUser.php`:

```php
<?php
if ($_SERVER["REQUEST_METHOD"] === "POST") {
    $firstName = trim($_POST["firstName"] ?? "");
    $familyName = trim($_POST["familyName"] ?? "");

    if ($firstName === "" || $familyName === "") {
        http_response_code(400);
        echo "Both names are required.";
    } else {
        $user = ["firstName" => $firstName, "familyName" => $familyName];
        echo htmlspecialchars($user["firstName"], ENT_QUOTES, "UTF-8");
    }
}
?>
```

- `$_POST` and `$_GET` are **superglobal arrays** accessible within PHP scopes.
- Keys match input `name` attributes, not necessarily their IDs.
- `?? ""` supplies a fallback if the key is absent.
- `trim` removes surrounding whitespace; check for empty required fields server-side.
- `htmlspecialchars` escapes text for HTML output. It is not SQL escaping.
- `var_dump($_POST)` is useful during development to inspect types and values, as on the slide; avoid exposing private submitted data in production output.

This compact example assumes single text values for these fields. A production handler should also reject unexpected field types and enforce length limits.

**Creating a PHP array does not save a user.** Database persistence needs a separate SQL operation. Use parameterised queries for user-supplied SQL values; this is supplementary guidance, not code supplied by this slide.

## 14. Runnable practice: understand submitted fields

**Supplementary Python model, not PHP or a live HTTP request.** This demonstrates how field names become keys. Predict the output before running it.

````python
from urllib.parse import urlencode, parse_qs
fields = {"firstName": "Ada", "familyName": "Lovelace"}
encoded = urlencode(fields)
received = parse_qs(encoded)
print("Form encoding:", encoded)
print("Parsed firstName:", received["firstName"][0])
assert received["firstName"] == ["Ada"]
assert received["familyName"] == ["Lovelace"]
# parse_qs returns lists; ordinary scalar PHP form fields are accessed differently.
````

## 15. Runnable practice: persistence behind a feature

**Supplementary SQL exercise using disposable in-memory SQLite, not a MySQL connection.** The portable SQL here demonstrates inserting a submitted user's data and retrieving it. SQLite `?` placeholders belong to Python's sqlite3 API; PHP driver APIs have their own binding calls. No external database or server is changed.

**Task:** before reading the solution cell, predict which row a search for `Ada` will return and why input values are passed separately from the SQL string.

````python
import sqlite3
with sqlite3.connect(":memory:") as connection:
    connection.execute("CREATE TABLE Users (user_id INTEGER PRIMARY KEY, first_name TEXT NOT NULL, family_name TEXT NOT NULL)")
    connection.execute("INSERT INTO Users VALUES (?, ?, ?)", (1, "Ada", "Lovelace"))
    connection.execute("INSERT INTO Users VALUES (?, ?, ?)", (2, "Grace", "Hopper"))
    rows = connection.execute(
        "SELECT user_id, first_name, family_name FROM Users WHERE first_name = ? ORDER BY user_id",
        ("Ada",),
    ).fetchall()
    print(rows)
    assert rows == [(1, "Ada", "Lovelace")]
# Values are bound separately so they are treated as data, not SQL syntax.
````
