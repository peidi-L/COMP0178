## 4. Database design: what the report must demonstrate (page 1)

- Follow the structured design methods taught in class.
- Translate the auction requirements into the data to store, operations/transactions to support, permitted changes, and access privileges.
- Create an **entity relationship (ER) diagram** showing relevant entities, attributes, and relationships.
- Translate the ER diagram systematically into a **database schema**: the tables, columns, and relationships used by the implementation.
- Show that the schema is in **third normal form (3NF)**.
- Explanation: an **entity** is something about which you store data; an **attribute** describes it; a **relationship** connects entities.
- Explanatory example: a User places a Bid on an Auction. Bid could store an amount and timestamp; Auction could store an end time and reserve price. These are candidate design choices, not a prescribed schema.
- A **primary key** identifies a row; a **foreign key** refers to a row in another table, helping represent relationships.
- A simplified 3NF intuition is to avoid storing unrelated or indirectly dependent facts together, reducing duplicated information and inconsistent updates.
- More precise supplementary definition: for every nontrivial functional dependency `X → A`, 3NF requires X to be a superkey or A to be a prime attribute (part of some candidate key).
- In the report, identify keys and functional dependencies and justify the normal form; merely stating "our database is in 3NF" does not provide the requested analysis.

## 5. Core functionality and its marks (pages 3-4)

- **Core 1 - Accounts and roles: 10%.** Users register and create accounts, with buyer/seller roles and different privileges.
- Explanation: roles determine what a user is allowed to do; displaying a role label alone does not demonstrate distinct privileges.
- **Core 2 - Auction creation: 10%.** Sellers create auctions with item description, category, starting price, reserve price, and end date.
- Explanation: the starting price is the initial bidding price; a reserve price is the seller's minimum acceptable sale price. Document how your system handles reserve conditions.
- **Core 3 - Search and browse: 15%.** Buyers search for types of items, browse categories, and visually rearrange listings within categories.
- Explanation: sorting listings by price or ending time is one possible interpretation of rearranging them; the brief does not prescribe particular controls.
- **Core 4 - Bidding and completion: 15%.** Buyers place bids and see other users' bids as received; the system manages the auction until its end time, awards the item to the highest bidder, and confirms the outcome to winner and seller.
- The brief does not prescribe a live-update transport, tie-breaking rule, or what happens when no bid meets the reserve. Record and implement consistent assumptions rather than treating these details as settled by the PDF.
- The four core components total **50% of the coursework marking allocation**.

## 6. Extra functionality and its marks (page 4)

- **E1-4 - Database-backed extensions of core features: up to 20%.** The brief gives a general category, not a fixed list of four required extensions.
- Extra features must relate to core functionality and involve database use; cosmetic changes alone do not meet that description.
- **E5 - Watching and email updates: 5%.** Buyers watch auctions and receive emailed updates on bids, including notifications when outbid.
- **E6 - Collaborative-filtering recommendations: up to 5%.** Recommend auctions based on the bidding activity of buyers with similar past interests.
- Explanatory example: if buyers with bidding histories similar to yours are now bidding on an item, the system may recommend it to you.
- Recommending all popular items is not necessarily collaborative filtering; explain how similarity or shared bidding behavior informs the recommendation.
- The listed extras total **30% of the coursework marking allocation**.

## 7. How the project is assessed (pages 3-4)

- **Functionality: 80%.** Core capabilities contribute 50%; extra capabilities contribute up to 30%.
- **Database design: 20%.** ER diagram contributes 10%; schema design and normalization contribute 10%.
- These percentages describe the **coursework**, which itself is worth 50% of the module.
- Before individual peer adjustment, database design therefore represents 10 percentage points of the overall module mark, and functionality represents 40.
- Marks consider the **quality and completeness** of each implemented capability, not just whether a feature is mentioned.
- The design table gives levels of **0, 5, 8, and 10 points**: absent; basic with flaws/misunderstandings; mostly correct with minor flaws; strong answer.
- For auction creation, the listed levels are **5**, **7-8**, and **10** points for partial, minor-issue, and strong implementation.
- For bidding/completion, the listed levels are **8**, **10-12**, and **15** points for the corresponding quality levels.
- E1-4 is awarded **0-20 points** depending on implementation; E6 is awarded **0-5 points**.
- **UI presentation/design and usability are not assessed.** The brief asks you to focus time on the database, queries, and application functionality rather than visual polish.
- Suggested priority: make the core capabilities demonstrable, ensure the schema/report are sound, and then add database-backed extras.

## 8. Group deliverable: demonstration video (page 2)

- Make a **5-8 minute video** demonstrating the working system with spoken voice-over narration.
- Demonstrate each achieved capability explicitly, in the **order listed in the Design Brief**.
- Refer to the relevant capability in the narration; preferably show the requirements list again during the video.
- Demonstrate through the website interface or by executing relevant database queries in **phpMyAdmin**.
- Extra functionality related to a core feature can be shown alongside it or in a separate section.
- Moodle contains example videos from previous years.
- Upload the video to **YouTube as unlisted**, then include its URL in the report.
- Explanation: unlisted means the video is accessed through its link rather than publicly listed; it is different from a private video requiring account-specific access.

## 9. Group deliverable: design report (page 2)

- Submit **one PDF file** through Moodle.
- Put the required elements in the following exact order.
- **1. YouTube video URL.** This gives the marker access to the demonstration.
- **2. ER diagram and assumptions.** Explain assumptions about the processes using the data.
- **3. Schema listing and explanation.** List table names and attributes and explain how they translate the ER diagram.
- **4. Third-normal-form analysis.** Show why the schema satisfies 3NF.
- **5. Database-query listing.** Give each query a very brief, high-level explanation of its purpose; preferably group queries by source file or supported functionality.
- The report does **not** need an introduction or conclusion.
- Explanatory example of a query description: "Retrieve open auctions in the selected category, ordered by ending time." Explain the purpose rather than narrating every SQL keyword.

## 10. Group deliverable: source-code archive (page 2)

- Submit an archive of the source files needed by the site, including relevant **PHP, SQL, HTML, and CSS** files.
- Include the **group number in the ZIP filename**.
- Submit source code only: exclude external/prewritten libraries or dependencies and large datasets such as large image folders.
- Distinguish using the starter's open-source tools during development from bundling third-party libraries in the submitted source archive.
- Suggested packaging check: ensure the database-creation SQL and application source are present, and that the archive can be understood independently of one person's local setup.

- The group mark is **adjusted using groupmates' assessments** to determine each person's final coursework mark.
- The brief does not state the numerical adjustment formula; use the agreed peer scheme and relevant Moodle guidance.
- Suggested working habit: record contributions and decisions throughout the project so assessments can reflect actual work.
