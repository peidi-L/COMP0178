## 2. Technology stack - pages 3-5

A **stack** is the collection of technologies that cooperate to run an application.

| Component | Role | What it is not |
|---|---|---|
| Browser | Displays HTML/CSS and runs browser JavaScript | The database server |
| Apache | Receives web requests and serves responses | The SQL database |
| PHP | Runs server-side application logic and database queries | Code executed by the browser |
| MySQL / MariaDB | Stores and queries relational data | A page-styling tool |
| phpMyAdmin | Browser-based administration interface for the database | The database engine itself |
| XAMPP | Bundles a local web-development environment | A programming language |
| Git / GitHub | Tracks changes / supports shared repository collaboration | A replacement for the running database |
| IDE or editor | Helps you read, edit, and debug code | The server executing PHP |

The lecture suggests XAMPP; alternatives mentioned include MAMP and WAMPserver. Component versions and local configuration can differ.

```text
Browser -> Apache/PHP -> MySQL or MariaDB
Browser <- generated response <- PHP query results
```

**Why localhost matters:** opening a PHP file directly as a local file does not ask a PHP-enabled server to execute it. Request it through your configured server URL, such as `http://localhost/example.php`; the exact port and directory depend on your installation.
